```python
# Ejercicio 1: Integración de una función
import numpy as np
from scipy.integrate import quad 
import time

ref = 2.55982 # Valor de referencia para la integral, obtenido analíticamente.
# Intervalo de integración
a = 0
b = 8
f = lambda x: np.exp(-0.4 * x) * (1 + 0.5 * np.sin(3 * x)) #Con un lambda function, se define la función a integrar.

# Se definen las tolerancias para detener el cálculo de la integral
tolerancia_abs = 1e-6 #Tolerancia entre el valor calculado y el valor que SciPy asume como el real
tolerancia_rel = 1e-6 # Tolerancia del error relativo entre el valor calculado y el valor real

# Se mide el tiempo de ejecución de la integración numérica usando scipy.integrate.quad
start_time = time.time()
resultado =  quad(f, a, b, epsabs=tolerancia_abs, epsrel=tolerancia_rel) # scipy.integrate.quad(f, a, b, epsabs=tolerancia_abs, epsrel=tolerancia_rel) devuelve una tupla con el valor de la integral y el error estimado.
end_time = time.time()
execution_time = end_time - start_time #Tiempo de ejecución en segundos

# Calculo de error absoluto y relativo
error_absoluto = abs(resultado[0] - ref)
error_relativo = error_absoluto / abs(ref) * 100

# Resultados
print (f"Resultado: {resultado[0]}")
print (f"Error estimado: {resultado[1]}")
print (f"Error absoluto: {error_absoluto}")
print (f"Error relativo: {error_relativo}%")
print (f"Tiempo de ejecución: {execution_time}")
```

```python
# Ejercicio 2: Integración de 50 datos equiespaciados
import numpy as np
from scipy.integrate import trapezoid, simpson # Se importan las funciones trapezoid y simpson de scipy.integrate
import matplotlib.pyplot as plt # Se importa pyplot para graficar los datos

data = np.loadtxt("datos_sensor.csv", delimiter=",", skiprows=1) # Se carga el archivo CSV con los datos del sensor, se omite la primera fila (encabezado) y se separan los valores por comas
x = data[:,0] # Se extrae la primera columna del archivo CSV, que corresponde a los valores de x tiempo
y = data[:,1] # Se extrae la segunda columna del archivo CSV, que corresponde a los valores de mediciones del sensor
I_trap = trapezoid(y, x) # Calculo con regla del trapecio
I_simp = simpson(y, x=x) # calculo con regla de Simpson

print (f"Integral usando el método del trapecio: {I_trap}")
print (f"Integral usando el método de Simpson: {I_simp}")

# Se grafica la función y los puntos de datos del sensor
plt.plot(x, y, 'o', label='Datos del sensor') # Se grafican los puntos de datos del sensor como círculos
plt.plot(x, y, label='Función del sensor') # Se grafican los puntos de datos del sensor como una línea
plt.xlabel('Tiempo (s)') # Se etiqueta el eje x como "Tiempo (s)"
plt.ylabel('Medición del sensor') # Se etiqueta el eje y como "Medición del sensor"
plt.legend()
plt.show()

#Diferencia entre ambos métodos de integración
diferencia = abs(I_trap - I_simp)
print (f"Diferencia entre ambos métodos de integración: {diferencia}")
```

```python
# Ejercicio 3: Diferenciación
import numpy as np
import pandas as pd

# Cargar el CSV (pandas ignora automáticamente los encabezados)
datos = pd.read_csv('datos_sensor.csv')
x = datos['x'].values
y = datos['y'].values

n = len(x)
h = x[1] - x[0]

# Inicializar los arreglos de derivadas con ceros
dy_adelante = np.zeros(n)
dy_central = np.zeros(n)

# Diferencia hacia adelante
# np.diff(y) calcula y[i+1] - y[i] para todo el arreglo
dy_adelante[:-1] = np.diff(y) / h

# Se hace diferencia hacia atrás para el extremo final
dy_adelante[-1] = (y[-1] - y[-2]) / h

# Diferencia central
# y[2:] toma desde el tercer elemento hasta el final (futuros)
# y[:-2] toma desde el primer elemento hasta el antepenúltimo (pasados)
dy_central[1:-1] = (y[2:] - y[:-2]) / (2 * h)

# Tratamiento de los extremos para diferencia central
dy_central[0] = (y[1] - y[0]) / h      # En el primer punto, se hace diferencia hacia adelante
dy_central[-1] = (y[-1] - y[-2]) / h   # En el último punto, se hace diferencia hacia atrás

# Resultados
print(f"{'i':<5} {'x':>12} {'y':>12} {'dy (Adelante)':>15} {'dy (Central)':>15}")
print("-" * 63)

for i in range(n):
    print(f"{i:<5} {x[i]:>12.6f} {y[i]:>12.6f} {dy_adelante[i]:>15.6f} {dy_central[i]:>15.6f}")
```
