```python
import numpy as np
import matplotlib.pyplot as plt
import os
import time
```

1.Se carga los datos del CSV generado por tu compañero

```python

data = np.loadtxt("trayectoria_robot.csv", delimiter=",", skiprows=1)
t, x, y = data[:,0], data[:,1], data[:,2]
```

2. Cálculos de cinemática diferencial usando la solución vectorizada de NumPy

```python
# Tiempo inicial de ejecución del algoritmo
start_time = time.time()
vx = np.gradient(x, t)
vy = np.gradient(y, t)
v = np.sqrt(vx**2 + vy**2)
theta = np.unwrap(np.arctan2(vy, vx))
omega = np.gradient(theta, t)
#Tiempo final de ejecución del algoritmo
end_time = time.time()
execution_time = end_time - start_time

print(f"Tiempo de ejecución del algoritmo: {execution_time:.6f} segundos")
```

3. Generación de las gráficas usando Matplotlib

```python
fig, axs = plt.subplots(2, 2, figsize=(10, 7))
fig.suptitle('Parte 2: Cinemática Diferencial con Python (NumPy)')

axs[0,0].plot(x, y, 'b.-', label='Trayectoria')
axs[0,0].set_title('Trayectoria y vs. x')
axs[0,0].set_xlabel('x (m)'); axs[0,0].set_ylabel('y (m)')
axs[0,0].grid(True)

axs[0,1].plot(t, v, 'g.-', label='Vel. Lineal')
axs[0,1].set_title('Velocidad Lineal v vs. t')
axs[0,1].set_xlabel('t (s)'); axs[0,1].set_ylabel('v (m/s)')
axs[0,1].grid(True)

axs[1,0].plot(t, theta, 'm.-', label='Orientación')
axs[1,0].set_title('Orientación θ vs. t')
axs[1,0].set_xlabel('t (s)'); axs[1,0].set_ylabel('θ (rad)')
axs[1,0].grid(True)

axs[1,1].plot(t, omega, 'r.-', label='Vel. Angular')
axs[1,1].set_title('Velocidad Angular ω vs. t')
axs[1,1].set_xlabel('t (s)'); axs[1,1].set_ylabel('ω (rad/s)')
axs[1,1].grid(True)

plt.tight_layout()
plt.show()
```
